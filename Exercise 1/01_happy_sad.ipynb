Feliz ou Triste?

Objetivo:

Classificar mensagens como:

feliz
triste

```python
X = [
    [1,0,1,1], # feliz
    [1,1,1,1], # feliz
    [1,1,0,1], # feliz
    [0,1,0,1], # feliz

    [0,0,1,0], # triste
    [1,0,0,0], # triste
    [0,0,0,0], # triste
    [0,1,1,0]  # triste
]

y = [
    1,
    1,
    1,
    1,

    0,
    0,
    0,
    0
]
```

```python
from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier()

modelo.fit(X, y)

print('Modelo treinado com sucesso!')
```

```python
# Testar com um novo dado
novo = [[1,1,1,0]]

resultado = modelo.predict(novo)

if resultado[0] == 1:
    print('Feliz')
else:
    print('Triste')

print(resultado)
```

```python
# Novos dados com arquivo csv
import pandas as pd
df = pd.read_csv('happy_sad.csv')
```

```python
# Exibindo dados
df.head()

#Cada linha = uma pessoa

#Cada coluna = uma característica

#A última coluna = resposta correta
```

```python
# Separação entre features (X) e target (y)
# Se não usar o df.drop:
# Você está entregando a resposta junto com a pergunta.

X = df.drop("sentimento", axis=1) # para remover a resposta das entradas.

y = df["sentimento"] # guardar a resposta separadamente.
```

```python
print(X.head()) # X = características da pessoa
```

```python
print(y.head()) # y = estado emocional da pessoa
```

```python
# Treinar
from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier()


# fit() significa aprender. O modelo está estudando os exemplos.
modelo.fit(X, y)

print('Modelo treinado!')
```

Pessoa nova:

* Sorriu? Sim
* Recebeu boa notícia? Sim
* Está cansado? Não
* Saiu com amigos? Sim

```python
# Nova pessoa
novo = [[1,1,0,1]]
```

```python
# Feliz ou triste?
nova_pessoa = modelo.predict(novo)

if nova_pessoa == 1:
  print('Essa pessoa é feliz!')
else:
  print('Essa pessoa é triste!')
```
