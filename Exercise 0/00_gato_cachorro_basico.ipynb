```python
# Ler aqui CSV
import pandas as pd

df = pd.read_csv('CatDog.csv')
```

```python
# Exibir registros.
df
```

```python
# Criar regra simples:
# Criar variável 'peso'.
# Usar o valor da coluna 'peso' do seu DataFrame 'df'.
# Vamos pegar o peso do primeiro registro como exemplo:
peso = df['peso'].iloc[0]

if peso < 10:
    print("Gato")
else:
    print("Cachorro")
```

```python
# Calcular acurácia manualmente.
from sklearn.metrics import accuracy_score

y_pred = [0, 2, 1, 3]
y_true = [0, 1, 2, 3]
accuracy_score(y_true, y_pred)
```

### Criando um Modelo de Classificação

Vamos criar um modelo de classificação usando a coluna `peso` como nossa característica (feature) para prever a `raca`.

```python
# 1. Definir as características (X) e o alvo (y)
# A característica será 'peso' e o alvo será 'raca'.
X = df[['peso']]
y = df['raca']

print("Características (X):")
print(X)
print("\nAlvo (y):")
print(y)
```

```python
# 2. Codificar a variável alvo (y)
# A coluna 'raca' é categórica ('gato', 'cachorro'),
# então precisamos convertê-la em valores numéricos.

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_encoded = le.fit_transform(y)

print("Alvo codificado (y_encoded):")
print(y_encoded)
print("\nMapeamento de classes:")
for i, label in enumerate(le.classes_):
    print(f"{label}: {i}")
```

```python
# 3. Escolher e treinar um modelo de classificação
# Usaremos um K-Nearest Neighbors Classifier, que é simples e eficaz para conjuntos de dados pequenos.
from sklearn.neighbors import KNeighborsClassifier

# Criar o modelo (com 1 vizinho, dado o tamanho do nosso dataset)
model = KNeighborsClassifier(n_neighbors=1)

# Treinar o modelo
model.fit(X, y_encoded)

print("Modelo KNeighborsClassifier treinado com sucesso!")
```

```python
# 4. Fazer previsões
# Vamos prever as raças para os dados existentes.
y_pred_encoded = model.predict(X)

# Decodificar as previsões para os nomes originais das raças
y_pred_decoded = le.inverse_transform(y_pred_encoded)

print("Previsões do modelo (decodificadas):")
print(y_pred_decoded)
print("\nValores reais:")
print(y.values)
```

```python
# 5. Avaliar a acurácia do modelo
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_encoded, y_pred_encoded)

print(f"Acurácia do modelo: {accuracy:.2f}")
```

Novas entradas para 'peso' e usar o modelo treinado para prever se elas são 'gato' (gato) ou 'cachorro' (cachorro).


```python
# Testar com novos dados
import pandas as pd

# Criar novos dados para teste
novos_pesos = pd.DataFrame({'peso': [3, 7, 25, 40]})

# Fazer previsões usando o modelo treinado
novas_predicoes_encoded = model.predict(novos_pesos)

# Decodificar as previsões para os nomes originais das raças
novas_predicoes_decodificadas = le.inverse_transform(novas_predicoes_encoded)

print("Novos dados de peso:")
print(novos_pesos)
print("\nPrevisões para os novos dados:")
print(novas_predicoes_decodificadas)
```

O modelo previu com sucesso os tipos de animais para os novos pesos que você forneceu! Para os pesos 3 e 7, previu 'gato' (gato) e, para os pesos 25 e 40, previu 'cachorro' (cachorro). Isso demonstra como o modelo treinado consegue classificar novos dados, nunca vistos antes, com base na variável 'peso'.
