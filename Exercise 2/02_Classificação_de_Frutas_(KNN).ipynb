Objetivo

Classificar uma fruta utilizando:

* Peso
* Diâmetro
* Cor

Fazer a leitura do dataset

```python
import pandas as pd

df = pd.read_csv("frutas.csv")

# Mostre as primeiras 3 linhas do dataset
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

cor_encoder = LabelEncoder()
df["Cor"] = cor_encoder.fit_transform(df["Cor"])

fruta_encoder = LabelEncoder()
df["Fruta"] = fruta_encoder.fit_transform(df["Fruta"])

df
```

### Entendendo o `LabelEncoder`

O `LabelEncoder` do `sklearn.preprocessing` é uma ferramenta usada para converter rótulos categóricos em valores numéricos. Isso é necessário porque a maioria dos algoritmos de aprendizado de máquina trabalha apenas com dados numéricos.

Neste código:

*   `encoder = LabelEncoder()`: Cria uma instância do `LabelEncoder`.
*   `df["Cor"] = encoder.fit_transform(df["Cor"])`:
    *   `fit_transform()`: Primeiro, o `LabelEncoder` *aprende* todos os valores únicos na coluna "Cor" (por exemplo, "Vermelho", "Laranja", etc.) e atribui um número a cada um (0, 1, 2...). Em seguida, ele *transforma* a coluna, substituindo os textos pelos números correspondentes.
*   `df["Fruta"] = encoder.fit_transform(df["Fruta"])`: O mesmo processo é aplicado à coluna "Fruta".

**Exemplo:** Se a coluna "Cor" tinha 'Vermelho', 'Laranja', 'Verde', o `LabelEncoder` poderia transformá-los em 0, 1, 2, respectivamente. Os números atribuídos dependem da ordem em que o `LabelEncoder` encontra os valores únicos.

```python
X = df[["Peso","Diametro","Cor"]]
y = df["Fruta"]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
    X,y,
    test_size=0.25, # 25% dos dados serão usados para teste, 75% para treino.
    random_state=42 # Garante que a divisão seja a mesma sempre que o código for executado, para reprodutibilidade.
)
```

### Por que dividir os dados em treino e teste (`train_test_split`)?

Dividir seu dataset em conjuntos de treino e teste é uma prática fundamental em aprendizado de máquina por várias razões:

1.  **Avaliação Realista do Modelo**: O principal objetivo é avaliar o desempenho do seu modelo em dados *não vistos* anteriormente. O modelo aprende padrões nos dados de treino. Se o avaliarmos nos mesmos dados de treino, ele pode parecer ter um desempenho excelente, mas isso não garante que ele funcionará bem com novos dados do mundo real. O conjunto de teste simula esses novos dados.

2.  **Detecção de Overfitting (Sobreajuste)**: O overfitting ocorre quando o modelo aprende os dados de treino tão bem que começa a memorizar o "ruído" ou as particularidades específicas desse conjunto, em vez de aprender os padrões gerais. Isso faz com que ele performe mal em dados novos. Ao usar um conjunto de teste separado, podemos identificar se o modelo está sobreajustado.

3.  **Generalização**: Um bom modelo deve ser capaz de generalizar, ou seja, de fazer previsões precisas em dados que nunca viu antes. O conjunto de teste nos dá uma medida confiável da capacidade de generalização do modelo.

Em resumo, a divisão em treino e teste garante que você tenha uma estimativa mais precisa de quão bem seu modelo funcionará na prática.

```python
from sklearn.neighbors import KNeighborsClassifier

modelo = KNeighborsClassifier(n_neighbors=3)

modelo.fit(X_train,y_train)
```

```python
predicoes = modelo.predict(X_test)

predicoes
```

```python
# O accuracy_score é uma métrica comum para avaliação de modelos de classificação.
# Ele calcula a proporção de previsões corretas (tanto verdadeiros positivos quanto verdadeiros negativos) em relação ao total de previsões.
# Um valor de 0.75 significa que o modelo acertou 75% das classificações no conjunto de teste.
from sklearn.metrics import accuracy_score

accuracy_score(y_test,predicoes)
```

Teste - Nova fruta

```python
# Definindo os valores para as características de uma nova fruta
peso = 165
diametro = 7.3
# Convertendo a cor 'Vermelho' para o seu valor numérico codificado usando o 'cor_encoder'.
# O '[0]' é usado para extrair o valor numérico do array retornado por 'transform'.
cor = cor_encoder.transform(["Vermelho"])[0]

# Usando o modelo treinado para prever a 'Fruta' com base nos novos valores de 'peso', 'diametro' e 'cor'.
# O '[[...]]' é necessário porque o método 'predict' espera uma entrada 2D (mesmo para uma única amostra).
modelo.predict([[peso,diametro,cor]])
```

O código previu array([2]). Isso significa que, com base nas características fornecidas (peso 165, diâmetro 7.3, e cor 'Vermelho' codificada), o modelo previu a fruta que corresponde ao rótulo numérico 2.

Para descobrir qual fruta é essa, podemos usar o fruta_encoder para decodificar esse número de volta para o nome da fruta.






```python
predicao_decodificada = fruta_encoder.inverse_transform(modelo.predict([[peso, diametro, cor]]))
display(f"A fruta prevista é: {predicao_decodificada[0]}")
```
